# **STAGE 1**

**Paddy Crop Analysis Using Python**






**Business Problem**



Paddy cultivation is influenced by various agricultural factors such as soil type, paddy variety, seed rate, fertilizer application, nursery management, and irrigation practices. Understanding how these factors affect crop yield is essential for improving agricultural productivity. However, due to the large volume of cultivation data, identifying meaningful patterns and relationships manually is difficult. This project aims to analyze the Paddy Crop dataset using Python to uncover trends, visualize important relationships, and identify the key factors influencing paddy yield. The insights gained from this analysis can support better farming practices and data-driven agricultural decision-making.

**Dataset Information**



Source : UCI Machine Learning Repository – Paddy Dataset (https://archive.ics.uci.edu/dataset/1186/paddy+dataset)

Location : Tamil Nadu, India

Year/Timeline : Agricultural survey data collected from paddy cultivation records (Dataset published in 2025)

Domain : Agriculture / Crop Production / Precision Agriculture

**Objective**



1. Consolidate the Paddy Crop dataset agricultural attributes such as paddy variety, soil type, seed rate, fertilizer application, irrigation, and crop yield for effective analysis.
2. Analyze the distribution of key cultivation factors, including paddy varieties, soil types, land area, and farming practices, to understand the overall characteristics of the dataset.
3. Compare crop yield across different paddy varieties and soil types to identify the combinations associated with higher agricultural productivity and lower productivity.
4. Investigate the relationship between cultivation factors such as Agriblock, Paddy Variety, Soil types, Nursery and paddy yield using statistical analysis and data visualization techniques.

**Expected Outcome**


* Clean and preprocess the Paddy Crop dataset for analysis.
* Explore the distribution of agricultural variables through descriptive statistics and visualizations.
* Identify relationships between cultivation practices and paddy yield.
* Compare crop yield across different paddy varieties and soil types.
* Detect important patterns and trends affecting paddy production.

* Generate meaningful insights to support efficient agricultural planning.

# **STAGE 2**

**Data Cleaning, Data Transformation and Feature Engineering**




**Handle duplicates**

```python
#Check the Duplicates

print("Duplicate Rows:", df.duplicated().sum())
duplicates = df[df.duplicated()]
duplicates

```

```python
#Drop the Duplicates
df = df.drop_duplicates()

df.shape
```

**Handle missing values (impute or drop)**

```python
#find the missing value

print(df.isna().sum())
mis = df.isna().sum()

if (mis > 0).any():
    print("\nYour dataset has missing values.")
else:
    print("\nYour dataset does not have any missing values.")

```

**Column name rename or dropping unwanted columns**

```python
# Rename the required columns
df.rename(columns={
    'Seedrate(in Kg)': 'Seed_Rate',
    'LP_Mainfield(in Tonnes)': 'LP_Mainfield',
    'Nursery area (Cents)': 'Nursery_Area',
    'LP_nurseryarea(in Tonnes)': 'LP_NurseryArea',
    'Pest_60Day(in ml)': 'Pest_60Day',
    '30DRain( in mm)': 'Rain_D1_D30',
    '30DAI(in mm)': 'Irrigation_D1_D30',
    '30_50DRain( in mm)': 'Rain_D31_D50',
    '30_50DAI(in mm)': 'Irrigation_D31_D50',
    '51_70DRain(in mm)': 'Rain_D51_D70',
    '51_70AI(in mm)': 'Irrigation_D51_D70',
    '71_105DRain(in mm)': 'Rain_D71_D105',
    '71_105DAI(in mm)': 'Irrigation_D71_D105',
    'Trash(in bundles)': 'Trash_Bundles',
    'Paddy yield(in Kg)': 'Paddy_Yield'
}, inplace=True)

print(df.columns) # Display updated column names
```

**Remove leading and trailing spaces from column names and column values**

```python

df.columns = df.columns.str.strip()

cols = [
    "Agriblock",
    "Variety",
    "Soil Types",
    "Nursery",
    "Wind Direction_D1_D30",
    "Wind Direction_D31_D60",
    "Wind Direction_D61_D90",
    "Wind Direction_D91_D120"
]

for col in cols:
    df[col] = df[col].astype(str).str.strip().str.title()

print("Extra spaces removed from Column Names and Column Values")
```

**Treat outliers if required(optional)**

```python
plt.figure(figsize=(6, 5))

sns.boxplot(y=df["Paddy_Yield"])

plt.title("Box Plot of Paddy Yield")
plt.ylabel("Paddy Yield (Kg)")

plt.show()
```

```python
# Remove Extra columns

df.drop (columns=[
    'Min temp_D1_D30', 'Max temp_D1_D30',
       'Min temp_D31_D60', 'Max temp_D31_D60', 'Min temp_D61_D90',
       'Max temp_D61_D90', 'Min temp_D91_D120', 'Max temp_D91_D120',
       'Inst Wind Speed_D1_D30(in Knots)', 'Inst Wind Speed_D31_D60(in Knots)',
       'Inst Wind Speed_D61_D90(in Knots)',
       'Inst Wind Speed_D91_D120(in Knots)', 'Wind Direction_D1_D30',
       'Wind Direction_D31_D60', 'Wind Direction_D61_D90',
       'Wind Direction_D91_D120', 'Relative Humidity_D1_D30',
       'Relative Humidity_D31_D60', 'Relative Humidity_D61_D90',
       'Relative Humidity_D91_D120',
], inplace=True)

print("Columns removed successfully.")
df.columns
```

**Convert data types (if needed)**

```python
# Convert categorical columns
cat_cols = [
    "Agriblock",
    "Variety",
    "Soil Types",
    "Nursery"
]

df[cat_cols] = df[cat_cols].astype("category")
df.info()
```

**Feature Engineering**

```python
# Total Rainfall

df["Total_Rainfall"] = (
    df["Rain_D1_D30"] +
    df["Rain_D31_D50"] +
    df["Rain_D51_D70"] +
    df["Rain_D71_D105"]
)

# Total Irrigation
df["Total_Irrigation"] = (
    df["Irrigation_D1_D30"] +
    df["Irrigation_D31_D50"] +
    df["Irrigation_D51_D70"] +
    df["Irrigation_D71_D105"]
)

# Total Fertilizer Applied
df["Total_Fertilizer"] = (
    df["DAP_20days"] +
    df["Urea_40Days"] +
    df["Potassh_50Days"] +
    df["Micronutrients_70Days"]
)

# Yield per Hectare

df["Yield_per_Hectare"] = (
    df["Paddy_Yield"] / df["Hectares"]
)
df.info()
```

# **STAGE 3**

 **Statistical analysis & Visualizations**



```python
# Measure of Central Tendency

mean = df["Paddy_Yield"].mean()
median = df["Paddy_Yield"].median()
mode = df["Paddy_Yield"].mode()[0]

print("Mean   :", mean)
print("Median :", median)
print("Mode   :", mode)
```

```python
# Chart - Visualization – Histogram with Mean & Median
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,5))

sns.histplot(df["Paddy_Yield"], bins=20, kde=True)

plt.axvline(mean,
            color="red",
            linestyle="--",
            linewidth=2,
            label=f"Mean = {mean:.2f}")

plt.axvline(median,
            color="green",
            linestyle="-",
            linewidth=2,
            label=f"Median = {median:.2f}")

plt.title("Distribution of Paddy Yield")
plt.xlabel("Paddy Yield (Kg)")
plt.ylabel("Frequency")
plt.legend()

plt.show()
```

**Interpretation**

The histogram indicates that the paddy yield is slightly left-skewed, as the mean (22,517.73 kg) is lower than the median (24,636.00 kg). Most observations are concentrated between 20,000 kg and 32,000 kg, indicating that the majority of farms achieved moderate to high yields. No significant outliers are observed, and the presence of multiple peaks suggests variations in yield due to different cultivation conditions such as variety, soil type, or agricultural practices.

```python
# Variance
variance = df["Paddy_Yield"].var()

# Standard Deviation
std_dev = df["Paddy_Yield"].std()

print("Variance:", variance)
print("Standard Deviation:", std_dev)
```

```python
mean = df["Paddy_Yield"].mean()
std = df["Paddy_Yield"].std()

plt.figure(figsize=(8,5))

sns.histplot(df["Paddy_Yield"], bins=20, kde=True)

plt.axvline(mean, color='red', linestyle='--', label='Mean')
plt.axvline(mean + std, color='green', linestyle=':', label='+1 Std Dev')
plt.axvline(mean - std, color='green', linestyle=':', label='-1 Std Dev')

plt.title("Paddy Yield Distribution with Standard Deviation")
plt.xlabel("Paddy Yield (Kg)")
plt.ylabel("Frequency")
plt.legend()

plt.show()
```

**Interpretation**

The histogram displays the distribution of paddy yield, with the red line representing the mean and the green dotted lines indicating one standard deviation above and below the mean. Most observations fall within this range, demonstrating the overall variability of paddy yield in the dataset.

```python
# Skewness
skewness = df["Paddy_Yield"].skew()

# Kurtosis
kurtosis = df["Paddy_Yield"].kurt()

print("Skewness :", skewness)
print("Kurtosis :", kurtosis)
```
